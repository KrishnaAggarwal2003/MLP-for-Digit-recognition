# MLP from scratch -- Using Numpy & Pandas

### This module tries to explain the working mechanism of a deep learning neural network and it's training process.
- This a MLP for digit recognition using mnist_train.csv as training dataset.
- There are 3 models having varying number of hidden layers -- 0,1 and 2
- In one cycle (epoch), all the training data is passed through the network, as one sample at a time.
-  The weights are initialized at the beginning of the training process, and keeps on updating until the training epochs are finished.
- The cost value is calculated for each epochs as an indication for evaluating the training process.

```python
import numpy as np
import pandas as pd
```

```python
#Q1 Extracting the data

def path(file_):
    data=pd.read_csv(file_)
    y = np.array(data['label'])
    y=y.reshape(-1,1)

    x=np.array(data)
    x=np.delete(x, 0 ,axis=1)

    

    print('Shape of pixel-matrix: \n', x.shape)
    print('\n Shape of data-label: \n',y.shape)
    
    return (
        x,
        y
    )
    
x,y =path('mnist_train.csv') 
print(x,'\n')
print(y)
```

```python
# Q2 Zero Hidden layer Neural Network

# Setting up parameters
x_normalized=x/255
weights_ = np.random.uniform(low=-0.1, high=0.1, size=(10, 784))
bias_ = np.ones((10,1))


def zero_layer_model(x_value, y_value, weight_matrix, bias_matrix, learning_rate, num_epochs):
    
 for number in range(num_epochs):   
    total_cost = 0  # Initialize the total cost for tracking across all samples
    
    for a in range(len(x_value)):
        input_layer = x_value[a]

        # True labels which contain accurate probability values
        true_label = np.zeros(10)
        true_label[y_value[a]] = 1
        true_label = true_label.reshape(-1, 1)

        # Calculate the result for the output layer
        result = np.dot(weight_matrix, input_layer.reshape(-1, 1)) + bias_matrix

        # Applying softmax function to the output
        softmax_probs = np.exp(result - np.max(result, axis=0, keepdims=True))
        predict_label = softmax_probs / np.sum(softmax_probs, axis=0, keepdims=True)

        # Gradient descent
        error = predict_label - true_label

        # Compute the cost for the current sample
        cost_function = -np.sum(true_label * np.log(predict_label))
        total_cost += cost_function  # Accumulate the cost

        # Update weights and bias
        weight_matrix -= learning_rate * np.dot(error, input_layer.reshape(1, -1))
        bias_matrix -= learning_rate * error

    print(f'Cost value for {number+1}st epoch: {total_cost / len(x_value)}')
 
 return weight_matrix , bias_matrix
     

updated_weights , updated_bias = zero_layer_model( x_normalized, y, weights_, bias_, 0.05, 50)
print(f'Updated weights: \n{updated_weights} \n Shape: {updated_weights.shape}')
print(f'Updated bias: \n{updated_bias} \n Shape: {updated_bias.shape}')
```

## Momentum in Neural Networks

Momentum is a technique that helps accelerate gradient descent in directions of consistent decrease while damping oscillations in directions with high curvature. It achieves this by incorporating a fraction of the previous weight update into the current update.

## How Momentum Works in Gradient Descent

### Without Momentum (Standard Gradient Descent)

**Update Rule:**

$$
w = w - \eta \cdot \nabla J(w)
$$

**Where:**
- $w$: Weights
- $\eta$: Learning rate
- $\nabla J(w)$: Gradient of the cost function \(J(w)\) with respect to \(w\)

---

### With Momentum

Momentum introduces a velocity term \(v\) that smoothens updates:

**Equations:**

$$
v = \beta \cdot v - \eta \cdot \nabla J(w)
$$

$$
w = w + v
$$

**Where:**
- $v$: Velocity (stores the previous updates)
- $\beta$: Momentum coefficient (commonly \(0.9\))
- $\eta$: Learning rate
- $\nabla J(w)$: Gradient of the cost function \(J(w)\) with respect to \(w\)

---

### Key Differences

- **Standard Gradient Descent:** Directly uses the gradient to update weights.
- **With Momentum:** Accumulates gradients (scaled by \(\beta\)) to smooth updates and prevent oscillations.

### Benefits of Momentum

1.  Accelerates convergence on steep slopes.
2.  Reduces oscillations in directions with high curvature.
3.  Helps escape shallow local minima or plateaus.

```python
#Q6 Adding momentum and ReLU function to the previous question

# Setting n-dim of hidden layer 01 as 10
n = 10
# Setting n-dim of hidden layer 02 as 10
m = 10

# Setting up the parameters
x_normalized = x/255

weights_01 = np.random.randn(n, 784) * np.sqrt(2 / 784)
weights_02 = np.random.randn(m, n) * np.sqrt(2 / n)
weights_03 = np.random.randn(10, m) * np.sqrt(2 / m)

bias_01 = np.ones((n, 1))
bias_02 = np.ones((m, 1))
bias_03 = np.ones((10, 1))

def ReLU(x):
    return np.maximum(0, x)

def ReLU_derivative(x):
    return np.where(x > 0, 1, 0)


def two_layer_model(x_value, y_value, weight_input, bias_layer01, weight_layer01, bias_layer02, weight_layer02, bias_output, learning_rate, momentum, epochs):
    # Initialize velocities to zero
    weight_input_velocity = np.zeros((n, 784))
    bias_layer01_velocity = np.zeros((n, 1))
    weight_layer01_velocity = np.zeros((m, n))
    bias_layer02_velocity = np.zeros((m, 1))
    weight_layer02_velocity = np.zeros((10, m))
    bias_output_velocity = np.zeros((10, 1))

    for epoch in range(epochs):
        total_cost = 0
        
        for a in range(len(x_value)):
            input_layer = x_value[a]

            # True labels which contain accurate prob. values
            true_label = np.zeros(10)
            true_label[y_value[a]] = 1
            true_label = true_label.reshape(-1, 1)

            result_01 = np.dot(weight_input, input_layer.reshape(-1, 1)) + bias_layer01

            # HIDDEN LAYER 01
            # ReLU activation
            ReLU_01 = ReLU(result_01)

            result_02 = np.dot(weight_layer01, ReLU_01) + bias_layer02

            # HIDDEN LAYER 02
            # ReLU activation
            ReLU_02 = ReLU(result_02)

            result_03 = np.dot(weight_layer02, ReLU_02) + bias_output

            # Applying softmax function to output result
            softmax_probs = np.exp(result_03 - np.max(result_03, axis=0, keepdims=True))
            predict_label = softmax_probs / np.sum(softmax_probs, axis=0, keepdims=True)

            # Gradient descent
            error = predict_label - true_label
            epsilon = 1e-12  # A very small number
            cost_function = -np.sum(true_label * np.log(predict_label + epsilon))
            total_cost+=cost_function

            # Backpropagation
            delta_output = error
            delta_hidden02 = np.dot(weight_layer02.T, delta_output) * ReLU_derivative(ReLU_02)
            delta_hidden01 = np.dot(weight_layer01.T, delta_hidden02) * ReLU_derivative(ReLU_01)

            # Weight and bias updates with momentum
            weight_input_velocity = momentum * weight_input_velocity - learning_rate * np.dot(delta_hidden01, input_layer.reshape(1, -1))
            bias_layer01_velocity = momentum * bias_layer01_velocity - learning_rate * delta_hidden01

            weight_layer01_velocity = momentum * weight_layer01_velocity - learning_rate * np.dot(delta_hidden02, ReLU_01.reshape(1, -1))
            bias_layer02_velocity = momentum * bias_layer02_velocity - learning_rate * delta_hidden02

            weight_layer02_velocity = momentum * weight_layer02_velocity - learning_rate * np.dot(delta_output, ReLU_02.reshape(1, -1))
            bias_output_velocity = momentum * bias_output_velocity - learning_rate * delta_output

            weight_input += weight_input_velocity
            bias_layer01 += bias_layer01_velocity

            weight_layer01 += weight_layer01_velocity
            bias_layer02 += bias_layer02_velocity

            weight_layer02 += weight_layer02_velocity
            bias_output += bias_output_velocity
        
        print(f'Cost value for {epoch} epoch: {total_cost / len(x_value)}')    

    # Returning the resulting weights after training
    return (
        weight_input,
        bias_layer01,
        weight_layer01,
        bias_layer02,
        weight_layer02,
        bias_output
    )

learning_rate = 0.05
momentum = 0.8
epochs = 50

resulting_weights = two_layer_model(x_normalized, y, weights_01, bias_01, weights_02, bias_02, weights_03, bias_03, learning_rate, momentum, epochs)

# Print the resulting weights
print('\nResulting Weights:\n')
print('Weight_Input: \n', resulting_weights[0])
print('\n Bias_Layer01: \n', resulting_weights[1])
print(' \n Weight_Layer01: \n', resulting_weights[2])
print('\n Bias_Layer02: \n', resulting_weights[3])
print('\n Weight_Layer02: \n', resulting_weights[4])
print('\n Bias_Output: \n', resulting_weights[5])
```
